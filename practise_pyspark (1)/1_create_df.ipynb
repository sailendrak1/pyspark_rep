```python

import pyspark.sql.functions as F

data2=[("G",4,2,None),("H",None,4,5)]

columns=["A","X","Y","Z"]

dfpiv=spark.createDataFrame(data=data2,schema=columns)

dfpiv.show()

cols_to_unpivot = ["x","y","z"]

num_cols=len(cols_to_unpivot)

stack_expr=", ".join([f"'{c}',`{c}`" for c in cols_to_unpivot])
unpivot_expr=f"stack({num_cols},{stack_expr}) as (B,C)"

df_unpivot=dfpiv.select("A",F.expr(unpivot_expr))
df_unpivot.show()
df_final=df_unpivot.filter("C is not null")

df_final.show()
```

```python

data=[(1,"Manish","India",3000),(2,"Harish","India",200000),(3,"Munish","India",3000),(30,"Munish","India",300),(4,"Harsha","USA",200000),(5,"Darsha","UK",3000),(6,"Darsha","USA",200000),(7,"Krish","UK",50000000),(8,"rishikesh","USA",200000)]

columns=["id","name","country","salary"]

datadf=spark.createDataFrame(data=data,schema=columns)

datadf.show()

from pyspark.sql.window import Window
from pyspark.sql.functions import rank
from pyspark.sql.functions import dense_rank
from pyspark.sql.functions import row_number,desc


wn = Window.partitionBy("country").orderBy(desc("salary"))
datadf.withColumn("densernk",dense_rank().over(wn)).withColumn("rank",rank().over(wn)).withColumn("rownum",rank().over(wn)).show()
```
